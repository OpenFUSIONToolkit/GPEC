```julia
using LinearAlgebra

function randmul(n)
    A = fill(rand(), n, n) ; B = fill(rand(),n ,n)
    return sum(A*B)

end 

function randmul!(C, A, B)

    fill!(A, rand()) ; fill!(B, rand())

    mul!(C,A,B)
    return sum(C)

end

    
```

```julia
n=100;

A= fill(rand(),n ,n)

B = copy(A); C = copy(A);

(randmul(n), randmul!(C,A,B))

```

```julia
using LinearAlgebra

function inner_sum(A, vs)
    t = zero(eltype(A))
    for v in vs
        t += inner(v, A, v)
    end
    return t 
end 
inner(v, A, w) = dot(v, A*w)
```

```julia
using OrdinaryDiffEq, StaticArrays

function lorentz_rule(u, p, t)
        σ, ρ, β = p
    x,y,z = u
    dx = σ*(y-x)
    dy = x*(ρ-z) - y
    dz = x*y - β*z
    return SVector(dx,dy,dz)
end
    
```

```julia
u₀ = SVector(10.0, 10.0, 10.0)
p₀ = [10, 28, 8/3]
```

```julia
prob = ODEProblem(lorentz_rule, u₀, (0.0, 100.0), p₀)
```

```julia
alg = Vern9 #9th adaptive solver
```

```julia
sol = solve(prob; alg = alg)
```

```julia
(sol.t[3], sol.u[3])
```

```julia
sol[1, 3]
```

```julia
using Plots

plot(sol[1, :], sol[2, :], sol[3, :])
```

```julia
sol(5.6)
```

```julia
function lorenz_rule(u, p, t)
    σ, ρ, β = p
    x, y, z = u
    dx = σ*(y - x)
    dy = x*(ρ - z) - y
    dz = x*y - β*z
    return SVector(dx, dy, dz)
end

# 2. Define parameters and initial state
u₀ = @SVector [1.0, 1.0, 1.0]
p₀ = [10.0, 28.0, 8/3]

# 3. Create the system (In v3.0, this uses 'CoupledODEs' or the updated constructor)
lorenz = CoupledODEs(lorenz_rule, u₀, p₀)

# 4. Get the trajectory
# In the newest version, it returns (dataset, time)
tr, tvec = trajectory(lorenz, 100.0; Δt = 0.05)
```
