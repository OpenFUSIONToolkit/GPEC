```julia
# Load in necessary packages
using Pkg
using LinearAlgebra
using HDF5
using Plots
using LaTeXStrings
```

```julia
path_to_file = "../examples/DIIID-like_kinetic_example/";
```

```julia
fmath5 = h5open(joinpath(path_to_file, "fmat_debug.h5"), "r") # Change path as needed
fmats_det = read(fmath5["fmat_det"])
psi = read(fmath5["psi"])
q = read(fmath5["q"])
fbarmat_det = read(fmath5["fbarmat_det"])
close(fmath5)
```

```julia
plot(q, abs.(fmats_det), xlabel=L"q", ylabel=L"|\det(F)|", title=L"$|\det(F)|$ vs $q$", yscale=:log10, legend=:false)
```

```julia
detfvpsi = plot(psi, abs.(fmats_det), xlabel=L"\psi", ylabel=L"\det(F)",
    title=L"$|F|$ vs $\psi$", yscale=:log10,
    legend=:false)
savefig(detfvpsi, "detfvpsi_plot.png")
detfvpsi
```

```julia
plot(psi, abs.(fbarmat_det), xlabel=L"\psi", ylabel=L"\det(\bar{F})",
    title=L"$|\bar{F}|$ vs $\psi$", yscale=:log10,
    legend=:false)
```

```julia
eh5 = h5open("../examples/DIIID-like_ideal_example/euler.h5", "r") # Change path as needed
mlow = read(eh5["info/mlow"])
xi_psi = read(eh5["integration/xi_psi"])
psifac = read(eh5["integration/psi"])
wt = read(eh5["vacuum/wt"])
crit = read(eh5["integration/crit"])
psio = read(eh5["equil/psio"])
et = read(eh5["vacuum/et"])
close(eh5)

# scale energy eigenvector matrices
chi1 = 2π*psio
wt = wt*(chi1*1e-3)
println("Done reading euler.h5")
```

```julia
p = plot()
for m in 1:5
    plot!(psifac, imag.(xi_psi[m - mlow + 1, 1, :]), label="m=$m")
end
xlabel!(L"\psi_N")
ylabel!(L"\mathrm{Im}(\xi_\psi)")
title!("Least Stable Eigenmode " * L"\xi_\psi" * " for " * "m=1-5")
```

```julia
p2 = plot()
for m in 1:5
    plot!(psifac, real.(xi_psi[m - mlow + 1, 1, :]), label="m=$m")
end
xlabel!(L"\psi_N")
ylabel!(L"\mathrm{Re}(\xi_\psi)")
title!("Least Stable Eigenmode " * L"\xi_\psi" * " for " * "m=1-5")
```

```julia
p3 = plot()
for m in 1:5
    plot!(psifac, abs.(xi_psi[m - mlow + 1, 1, :]), label="m=$m")
end
xlabel!(L"\psi_N")
ylabel!(L"\mathrm{|\xi_\psi|}")
title!("Least Stable Eigenmode " * L"\xi_\psi" * " for " * "m=1-5")
```
