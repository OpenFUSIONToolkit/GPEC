```julia
# Load in necessary packages
using Pkg
using LinearAlgebra
using HDF5
using Plots
```

```julia
fmath5 = h5open("../examples/DIIID-like_ideal_example/fmat_debug.h5", "r") # Change path as needed
fmats_det = read(fmath5["fmat_det"])
psi = read(fmath5["psi"])
q = read(fmath5["q"])
fbarmat_det = read(fmath5["fbarmat_det"])
close(fmath5)
```

```julia
plot(q, abs.(fmats_det), xlabel="q", ylabel="|det(F)|", title="F Matrix Determinant vs q", yscale=:log10, legend=:false)
```

```julia
plot(psi, abs.(fmats_det), xlabel="Psi", ylabel="|det(F)|",
    title="F Matrix Determinant vs Psi", yscale=:log10,
    legend=:false)
```

```julia
plot(psi, abs.(fbarmat_det), xlabel="Psi", ylabel="|det(F̄_bar)|",
    title="F̄_bar Matrix Determinant vs Psi", yscale=:log10,
    legend=:false)
```
